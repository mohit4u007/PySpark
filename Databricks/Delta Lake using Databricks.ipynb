## Data Skipping and Z-Ordering in Delta Lake Tables | Optimize & Data Compaction Delta Lake Tables

```python
# Dataset
dbutils.fs.ls("/Volumes/samples/databricks/datasets/definitive-guide/data/retail-data/all/online-retail-dataset.csv")
```

```python
display(dbutils.fs.head("/Volumes/samples/databricks/datasets/definitive-guide/data/retail-data/all/online-retail-dataset.csv"))
```

```python
# write the data in the form of delta table
df = spark.read.csv("/Volumes/samples/databricks/datasets/definitive-guide/data/retail-data/all/online-retail-dataset.csv", header=True, inferSchema=True)
df.repartition(16).write.format("delta").mode("overwrite").option("path","/Volumes/workspace/default/data/output/sales_delta_z").save()
```

```python
df.write.mode("overwrite").saveAsTable("workspace.default.sales_delta_z")
# spark.sql("DROP TABLE IF EXISTS workspace.default.sales_delta_z")
```

```python
dbutils.fs.ls("/Volumes/workspace/default/data/output/sales_delta_z")
```

```python
display(spark.sql("SELECT * FROM workspace.default.sales_delta_z"))
```

```python
display(spark.sql("SELECT * FROM workspace.default.sales_delta_z WHERE InvoiceNo = '540646'"))
```

```python
display(spark.sql("SELECT MIN(InvoiceNo) AS MIN_INVOICE_NO, MAX(InvoiceNo) AS MAX_INVOICE_NO, _metadata.file_name FROM workspace.default.sales_delta_z GROUP BY _metadata.file_name ORDER BY MIN_INVOICE_NO"))
```

```python
# spark.conf.set("spark.databricks.delta.optimize.maxFileSize", 64*1024*8) this won't work so try the following
spark.sql("""
  ALTER TABLE workspace.default.sales_delta_z
  SET TBLPROPERTIES ('delta.targetFileSize' = '64mb')
""")

spark.sql("OPTIMIZE workspace.default.sales_delta_z ZORDER BY (InvoiceNo)")
```

```python
display(spark.sql("SELECT MIN(InvoiceNo) AS MIN_INVOICE_NO, MAX(InvoiceNo) AS MAX_INVOICE_NO, _metadata.file_name FROM workspace.default.sales_delta_z GROUP BY _metadata.file_name ORDER BY MIN(InvoiceNo)"))
```

## Delta Tables - Deletion Vectors and Liquid Clustering | Optimize Delta Tables | Delta Clustering

```python
# Read csv data from location "dbfs:/Volumes/samples/databricks/datasets/online_retail/data-001/data.csv"
display(spark.sql("SELECT * FROM read_files('dbfs:/Volumes/samples/databricks/datasets/online_retail/data-001/data.csv', header => true, format => 'csv')"))
```

```python
spark.sql("CREATE TABLE workspace.default.online_retail_sales AS SELECT * FROM read_files('dbfs:/Volumes/samples/databricks/datasets/online_retail/data-001/data.csv', header => true, format => 'csv')")
```

```python
display(spark.sql("SELECT * FROM workspace.default.online_retail_sales"))
```

```python
display(spark.sql("DESCRIBE EXTENDED workspace.default.online_retail_sales"))
```

```python
# Delete deletionvector
spark.sql("ALTER TABLE workspace.default.online_retail_sales SET TBLPROPERTIES ('delta.enableDeletionVectors' = 'false')")
```

```python
# Delete data without deletion vectors with InvoiceNo = '536365'
spark.sql("DELETE FROM workspace.default.online_retail_sales WHERE InvoiceNo = '536365'")
```

```python
display(spark.sql("DESCRIBE HISTORY workspace.default.online_retail_sales"))
```

```python
# Enable deletion vectors
spark.sql("ALTER TABLE workspace.default.online_retail_sales SET TBLPROPERTIES ('delta.enableDeletionVectors' = 'true')")
```

```python
# Delete data with deletionvector with InvoiceNo = '536367'
spark.sql("DELETE FROM workspace.default.online_retail_sales WHERE InvoiceNo = '536367'")
```

```python
display(spark.sql("DESCRIBE HISTORY workspace.default.online_retail_sales"))
```

```python
# spark.sql("OPTIMIZE workspace.default.online_retail_sales")
```

```python
# Liquid Clustering
spark.sql("ALTER TABLE workspace.default.online_retail_sales CLUSTER BY (InvoiceNo)")
```

```python
display(spark.sql("DESCRIBE HISTORY workspace.default.online_retail_sales"))
```
